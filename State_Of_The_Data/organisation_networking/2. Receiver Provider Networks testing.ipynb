# Setup

Notebook to troubleshoot receiver provider networks.

```
import pandas as pd
import numpy as np
from pyvis import network as net
from IPython.display import HTML
```

## Load data

```
data = pd.read_csv("outputs/all_links_list.csv")
```

```
data
```

# Filtered Plot

```
org = 'Palladium International Ltd (UK)'
```

```
#level 1
prov = data[data['Provider'].isin([org])].copy()
prov['colour'] = '#AA9BEF'

rec = data[data['Receiver'].isin([org])].copy()
rec['colour'] = '#DC7EAC'
```

```
#level 2
prov2 = data[data['Provider'].isin(list(prov['Receiver']))].copy()
prov2['colour'] = '#AA9BEF'

rec2 = data[data['Receiver'].isin(list(rec['Provider']))].copy()
rec2['colour'] = '#DC7EAC'
```

```
plot = pd.concat([prov,rec,prov2,rec2]).drop_duplicates()
```

```
#set up network
g=net.Network(height='900px', width='1200px',heading='',directed =True)
```

```
#list unique nodes, add to graph
nodes = pd.concat([plot[plot['Provider'] != org]['Provider'],
                        plot[plot['Receiver'] != org]['Receiver']]).drop_duplicates()

g.add_nodes(nodes)
g.add_node(org,color ='#785EF0')
```

```
#add edges
sources = plot['Provider']
targets = plot['Receiver']
colour = plot['colour']

edge_data = zip(sources, targets,colour)

for e in edge_data:
                src = e[0]
                dst = e[1]
                c = e[2]
                g.add_edge(src, dst, color =c)
```

```
g.force_atlas_2based(gravity=-50,
            central_gravity=0.01,
            spring_length=100,
            spring_strength=0.08,
            damping=0.4,
            overlap=1)
```

```
g.save_graph("outputs/simple_graph.html")
```

```
HTML(filename="outputs/simple_graph.html")
```
