# Which refs are removed?

Looking into filtered references that don't make the list.

```
import pandas as pd
import numpy as np
```

```
#read in final list from chords
filtered_orgs =  pd.read_csv("outputs/filtered_orgs.csv")

#read in all list with source
full_list = pd.read_csv("outputs/full_list.csv")
```

## So orgs can look at their removed references - and why!

```
reportingorg_ref = 'XM-DAC-3-1'
```

```
#filter to org of interest
org_list = full_list[full_list['reportingorg_ref'] == reportingorg_ref].drop_duplicates()
org_list
```

```
#is the org-id in the final list?
def compare(row):
    if row['org-id'] in (filtered_orgs['org-id']).values:
        return "valid"
    else:
        return "invalid"
```

```
org_list["filtered"] = org_list.apply(compare, axis=1)
org_list
```
