# Setup

Notebook to create one big org file from "the list".

```
import pandas as pd
import numpy as np
from datetime import datetime
```

```
today = datetime.today()
```

```
orglist = pd.read_csv("outputs/reporgs.csv")
orglist
```

# Create dataframe

```
organisations = {
    "organisation-identifier": orglist['org-id'],
    "@xml:lang": 'en',
	"@last-updated-datetime": today,
    "reporting-org/@ref": "GB-COH-09506232",
    "reporting-org/@type":'70',
    "reporting-org/narrative":"Open Data Services Co-operative",
    "name/0/narrative/0": orglist['Most Common Narrative'],
    "odsc:narrative-source":orglist['Source']
}

organisations_iati = pd.DataFrame(organisations)
organisations_iati
```

# Meta sheet

```
meta = {'#' : ['@version','@generated-datetime','@xmlns:cove','@xmlns:odsc'],
     'XMLRootTag iati-organisations' : ['2.03',today,'https://iati.cove.opendataservices.coop/','https://opendataservices.coop/'],
     'RootListPath iati-organisation':['','','',''],
     'IDName organisation-identifier':['','','',''],	
     'hashComments':['','','',''],
     }

meta_df = pd.DataFrame(meta)
meta_df
```

## Save final files

Removes previous output and saves new sheets.

```
!rm -rf outputs/organisations.xlsx

output_name = "outputs/organisations.xlsx"

with pd.ExcelWriter(output_name) as writer:
    meta_df.to_excel(writer, sheet_name="Meta", index=False)

    def add_sheet(dataframe, sheetname):
        dataframe.to_excel(writer, sheet_name=sheetname, index=False)

    add_sheet(organisations_iati, "Organisations")
```
